```python
import matplotlib.pyplot as plt
t = pd.read_csv(root/"results/tables/trajectories.csv")
z = t[(t.scenario=="medium") & (t.code=="B6")]
plt.figure(figsize=(8,4))
plt.plot(z.year,100*z.mvi_gdp,label="MVI/GDP")
plt.plot(z.year,100*z.use_dividend_gdp,label="User dividends/GDP")
plt.xlabel("Model years")
plt.ylabel("Percent of GDP")
plt.legend()
plt.tight_layout()
```
